````python
#1
def to_num(s):
    return float(str(s).replace(" ", "").strip())

def clean_city(s):
    city = str(s).strip().lower()
    
    cities = {
        "almaty": "Алматы",
        "astana": "Астана",
        "shymkent": "Шымкент",
        "karaganda": "Караганда",
        "aktobe": "Актобе",
        "atyrau": "Атырау"
    }

    return cities.get(city, city.capitalize())
````

````python
#3
revenue = {}
counts = {}

for o in orders:
    if o["status"] == "оплачен":
        c = o["category"]
        a = to_num(o["amount"])

        if c not in revenue:
            revenue[c] = 0
            counts[c] = 0

        revenue[c] += a
        counts[c] += 1

for c in revenue:
    print(f"{c}: выручка = {revenue[c]}, средний чек = {revenue[c] / counts[c]:.2f}")
````
